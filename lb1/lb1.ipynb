## Задание 1

Посмотреть документацию к API GitHub, разобраться как вывести список репозиториев для конкретного пользователя, сохранить JSON-вывод в файле *.json.


```python
import json
import requests

username = "LaintAge"
url = f"https://api.github.com/users/{username}/repos"

response = requests.get(url)

if response.status_code == 200:
    repos_data = response.json()

    if not repos_data:
        print("Нет публичных репозиториев")
    else:
        print(f"{'№':<4} {'Название':<35} {'Ссылка'}")
        print("-" * 110)

        for i, repo in enumerate(repos_data, start=1):
            name = repo.get("name", "")[:33]
            html_url = repo.get("html_url", "")
            print(f"{i:<4} {name:<35} {html_url}")

        print("-" * 110)
        print(f"Всего репозиториев: {len(repos_data)}")

    with open("github_repos.json", "w", encoding="utf-8") as f:
        json.dump(repos_data, f, indent=4, ensure_ascii=False)

    print("Данные сохранены в файл github_repos.json")
else:
    print(f"Ошибка {response.status_code}")
    print(response.text)
```

```python
TOKEN = ""
```

```python
import requests, json

r = requests.get(
    "https://www.googleapis.com/blogger/v3/users/self/blogs",
    headers={"Authorization": f"Bearer {TOKEN}"}
)

data = r.json()

with open("blogger_blogs.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
print("Сохранено в blogger_blogs.json")

for blog in data.get("items", []):
    print(f"{blog['name']} — {blog['url']}")
```
